```python
# question-1 Ans
from sklearn.datasets import load_wine
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load dataset
X, y = load_wine(return_X_y=True, as_frame=True)

# (a) Print the shape of X and the count of each class in y
print("--- (a) Dataset Info ---")
print("Shape of X:", X.shape)
print("Class counts in y:\n", y.value_counts().sort_index(), "\n")

# (b) Split into train and test with test_size=0.3, random_state=42
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("--- (b) Train/Test Shapes ---")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape, "\n")

# (c) Scale features (fit_transform on train, transform on test)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# (d) Train LogisticRegression and evaluate
MODD = LogisticRegression(max_iter=1000)
MODD.fit(X_train_scaled, y_train)
y_pred=MODD.predict(X_test_scaled)
MODD_acc=accuracy_score(y_test,y_pred)
print("--- (d) Model Evaluation ---")
print(f"Test Accuracy: {MODD_acc:.4f}\n")
print("Classification Report:\n", classification_report(y_test, y_pred))

```

```python
# question-2 Ans
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# Load and split data (unscaled)
X, y = load_wine(return_X_y=True, as_frame=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# (a) Train DecisionTreeClassifier
DTC = DecisionTreeClassifier(random_state=42)
DTC.fit(X_train, y_train)
DTC_pred=DTC.predict(X_test)
DTC_acc = accuracy_score(y_test, DTC_pred)
print(f"(a) Decision Tree Test Accuracy: {DTC_acc:.4f}")

# (b) Train RandomForestClassifier
RTC = RandomForestClassifier(n_estimators=100, random_state=42)
RTC.fit(X_train, y_train)
RTC_pred=RTC.predict(X_test)
RTC_acc = accuracy_score(y_test, RTC_pred)
print(f"(b) Random Forest Test Accuracy: {RTC_acc:.4f}\n")

# (c) Top 3 features by importance
importances = pd.Series(RTC.feature_importances_, index=X.columns).sort_values(
    ascending=False
)
print("(c) Top 3 Most Important Features:")
for feature, score in importances.head(3).items():
    print(f"{feature}: {score:.4f}")
print("(d) one sentence: why does the forest usualy score higher than the single tree?: Ans- A random forest usually scores higher than a single decision tree because combining predictions across many diverse trees reduces overfitting and cancels out individual tree errors.")
```

```python
#question-3 Ans

from sklearn.datasets import load_wine
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

# Load and split the dataset
X, y = load_wine(return_X_y=True, as_frame=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# (a) Train on unscaled training data
KNN_unscaled = KNeighborsClassifier(n_neighbors=5)
KNN_unscaled.fit(X_train, y_train)
KNN_pred =KNN_unscaled.predict(X_test)
KNN_acc_unscaled = accuracy_score(y_test,KNN_pred)
print(f"(a) Unscaled Test Accuracy: {KNN_acc_unscaled:.4f}")

# (b) Train on scaled training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

KNN_scaled = KNeighborsClassifier(n_neighbors=5)
KNN_scaled.fit(X_train_scaled, y_train)
KNN_pred= KNN_scaled.predict(X_test_scaled)
acc_scaled = accuracy_score(y_test, KNN_pred)
print(f"(b) Scaled Test Accuracy:   {acc_scaled:.4f}")

# (c) Print both numbers together on one line
print(f"\n(c) Unscaled Accuracy: {KNN_acc_unscaled:.4f} | Scaled Accuracy: {acc_scaled:.4f}")

print("\n(d) One sentence: scaling changed the KNN result a lot but barely moved the decision tree in Question 2,Why?: Ans-KNN relies on straight line distance, while Decision Trees spit one feature at a time using simle threshold rules that ignore overall scale.")
```

```python
# Question-4 Ans
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler

# Load the full Wine dataset (only X, completely ignoring y)
wine = load_wine()
X = pd.DataFrame(wine.data, columns=wine.feature_names)

# (a) Scale all of X with StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# (b) Run KMeans(n_clusters=3, n_init=10)
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
labels = kmeans.fit_predict(X_scaled)

# (c) Print how many rows landed in each cluster
df = X.copy()
df["cluster"] = labels

print("Number of rows in each cluster:")
print(df["cluster"].value_counts().sort_index())

# (d) Print the mean of alcohol, proline, and color_intensity for each cluster
print("\nMean feature values per cluster:")
print(df.groupby("cluster")[["alcohol", "proline", "color_intensity"]].mean().round(2))
```

```python
#question-5 Ans
import csv
import json

csv_file = """name,marks,city
Asha,78,Bangalore
Ravi,,Chennai
Meena,91,Bangalore
Imran,64,Hyderabad
Divya,88,"""

with open("students.csv", "w", newline="") as f:
    f.write(csv_file)

def read_marks(path):
    try:
        with open(path, mode="r", newline="") as f:
            reader = csv.DictReader(f)
            total_rows = 0
            valid_marks = []

            for row in reader:
                total_rows += 1
                marks_str = row.get("marks", "").strip()
                if marks_str:
                    valid_marks.append(float(marks_str))

            avg_marks = sum(valid_marks) / len(valid_marks) if valid_marks else 0
            summary = {
                "total_rows": total_rows,
                "rows_with_marks": len(valid_marks),
                "average_marks": avg_marks
            }

            with open("summary.json", "w") as out_f:
                json.dump(summary, out_f, indent=4)

            with open("summary.json", "r") as in_f:
                print(in_f.read())

            return summary

    except FileNotFoundError:
        print(f"File not found: {path}")
        return None

read_marks("students.csv")
read_marks("missing.csv")
```
